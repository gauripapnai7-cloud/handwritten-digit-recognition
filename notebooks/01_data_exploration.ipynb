```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow:", tf.__version__)
print("NumPy:", np.__version__)
print("All libraries imported successfully!")
```

```python
# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = (
    tf.keras.datasets.mnist.load_data()
)

print("Training images:", X_train.shape)
print("Training labels:", y_train.shape)

print("Testing images:", X_test.shape)
print("Testing labels:", y_test.shape)
```

```python
plt.figure(figsize=(12, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("Unique digits:", np.unique(y_train))
print("Minimum pixel value:", X_train.min())
print("Maximum pixel value:", X_train.max())
print("Image dimensions:", X_train[0].shape)
```

```python
digits, counts = np.unique(y_train, return_counts=True)

for digit, count in zip(digits, counts):
    print(f"Digit {digit}: {count} images")
```

```python
plt.figure(figsize=(10, 5))

plt.bar(digits, counts)
plt.xticks(range(10))
plt.xlabel("Digit")
plt.ylabel("Number of Images")
plt.title("MNIST Training Dataset - Digit Distribution")

plt.show()
```

```python
# Create a figure to display one MNIST image
plt.figure(figsize=(6, 6))

# Display the first training image in grayscale
plt.imshow(X_train[0], cmap="gray")

# Show the correct digit label as the title
plt.title(f"Digit: {y_train[0]}")

# Display a color bar representing pixel intensity values
plt.colorbar()

# Render the image
plt.show()
```
