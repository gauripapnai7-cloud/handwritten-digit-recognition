```python
# Import the required libraries
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

# Load the MNIST training and testing datasets
(X_train, y_train), (X_test, y_test) = (
    tf.keras.datasets.mnist.load_data()
)

# Display the original dataset shapes
print("Original training images:", X_train.shape)
print("Original testing images:", X_test.shape)
```

```python
# Convert pixel values from the range 0-255 to 0-1
# This helps the neural network learn more effectively
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Check the new pixel value range
print("Minimum pixel value:", X_train.min())
print("Maximum pixel value:", X_train.max())

# Check the data type after conversion
print("Data type:", X_train.dtype)
```

```python
# Add a channel dimension to each training image
# The value 1 represents one grayscale channel
X_train = np.expand_dims(X_train, axis=-1)

# Add a channel dimension to each testing image
X_test = np.expand_dims(X_test, axis=-1)

# Display the updated image shapes
print("Training images shape:", X_train.shape)
print("Testing images shape:", X_test.shape)
```

```python
# Import the function used to split the dataset
from sklearn.model_selection import train_test_split

# Split the original training data into training and validation sets
# Keep 10% of the original training data for validation
# stratify=y_train maintains similar digit proportions in both sets
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.10,
    random_state=42,
    stratify=y_train
)

# Display the sizes of all three datasets
print("Training images:", X_train.shape)
print("Validation images:", X_val.shape)
print("Testing images:", X_test.shape)

# Display the corresponding label shapes
print("Training labels:", y_train.shape)
print("Validation labels:", y_val.shape)
print("Testing labels:", y_test.shape)
```

```python
# Create a folder for processed data if it does not exist
os.makedirs("../data/processed", exist_ok=True)

# Save the training, validation, and testing data in a compressed file
np.savez_compressed(
    "../data/processed/mnist_preprocessed.npz",
    X_train=X_train,
    y_train=y_train,
    X_val=X_val,
    y_val=y_val,
    X_test=X_test,
    y_test=y_test
)

# Confirm that the file has been saved
print("Preprocessed MNIST data saved successfully!")
```

```python
# Load the saved preprocessed dataset
data = np.load("../data/processed/mnist_preprocessed.npz")

# Display the saved arrays and their shapes
for name in data.files:
    print(f"{name}: {data[name].shape}")

# Display the minimum and maximum normalized pixel values
print("Normalized pixel range:",
      data["X_train"].min(),
      "to",
      data["X_train"].max())
```
